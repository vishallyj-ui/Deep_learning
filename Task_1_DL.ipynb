```python
# ==========================================================
# TASK 1
# Vectorized MLP Forward and Backward Pass
# ==========================================================

# ----------------------------------------------------------
# Step 1 - Import Libraries
# ----------------------------------------------------------

import numpy as np

print("Libraries Imported Successfully")


# ----------------------------------------------------------
# Step 2 - Create Dataset
# ----------------------------------------------------------

np.random.seed(42)

X = np.random.randn(100, 4)

y = np.random.randint(0, 2, (100, 1))

print("Input Shape :", X.shape)
print("Target Shape:", y.shape)


# ----------------------------------------------------------
# Step 3 - Activation Functions
# ----------------------------------------------------------

def sigmoid(x):

    return 1 / (1 + np.exp(-x))


def sigmoid_derivative(x):

    s = sigmoid(x)

    return s * (1 - s)


# ----------------------------------------------------------
# Step 4 - Initialize Weights
# ----------------------------------------------------------

input_size = 4
hidden_size = 8
output_size = 1

W1 = np.random.randn(input_size, hidden_size) * 0.1
b1 = np.zeros((1, hidden_size))

W2 = np.random.randn(hidden_size, output_size) * 0.1
b2 = np.zeros((1, output_size))

print("Weights Initialized")


# ----------------------------------------------------------
# Step 5 - Forward Pass
# ----------------------------------------------------------

def forward(X):

    Z1 = np.dot(X, W1) + b1

    A1 = sigmoid(Z1)

    Z2 = np.dot(A1, W2) + b2

    A2 = sigmoid(Z2)

    return Z1, A1, Z2, A2


# ----------------------------------------------------------
# Step 6 - Calculate Loss
# ----------------------------------------------------------

def calculate_loss(y, prediction):

    loss = -np.mean(
        y * np.log(prediction + 1e-8) +
        (1 - y) * np.log(1 - prediction + 1e-8)
    )

    return loss


# ----------------------------------------------------------
# Step 7 - Backward Pass
# ----------------------------------------------------------

def backward(X, y, Z1, A1, Z2, A2):

    m = X.shape[0]

    # Output layer gradient
    dZ2 = A2 - y

    dW2 = np.dot(A1.T, dZ2) / m

    db2 = np.sum(dZ2, axis=0, keepdims=True) / m

    # Hidden layer gradient
    dA1 = np.dot(dZ2, W2.T)

    dZ1 = dA1 * sigmoid_derivative(Z1)

    dW1 = np.dot(X.T, dZ1) / m

    db1 = np.sum(dZ1, axis=0, keepdims=True) / m

    return dW1, db1, dW2, db2


# ----------------------------------------------------------
# Step 8 - Train the Network
# ----------------------------------------------------------

learning_rate = 0.5
epochs = 500

for epoch in range(epochs):

    # Forward
    Z1, A1, Z2, A2 = forward(X)

    # Loss
    loss = calculate_loss(y, A2)

    # Backward
    dW1, db1, dW2, db2 = backward(
        X,
        y,
        Z1,
        A1,
        Z2,
        A2
    )

    # Update weights
    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1

    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2

    if (epoch + 1) % 100 == 0:

        print(
            "Epoch",
            epoch + 1,
            "| Loss =",
            round(loss, 4)
        )


# ----------------------------------------------------------
# Step 9 - Make Predictions
# ----------------------------------------------------------

_, _, _, predictions = forward(X)

predicted_class = (predictions >= 0.5).astype(int)


# ----------------------------------------------------------
# Step 10 - Calculate Accuracy
# ----------------------------------------------------------

accuracy = np.mean(predicted_class == y)

print()
print("Training Completed")
print("Accuracy =", round(accuracy * 100, 2), "%")
```
