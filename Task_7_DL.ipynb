```python
# ==========================================================
# TASK 7
# ResNet Bottleneck Block with Grouped Convolution
# ==========================================================

# ----------------------------------------------------------
# Step 1 - Import Libraries
# ----------------------------------------------------------

import torch
import torch.nn as nn

print("Libraries Imported Successfully")


# ----------------------------------------------------------
# Step 2 - Create Input Data
# ----------------------------------------------------------

torch.manual_seed(42)

X = torch.randn(
    2,
    32,
    32,
    32
)

print("Input Shape:", X.shape)


# ----------------------------------------------------------
# Step 3 - Create Standard ResNet Block
# ----------------------------------------------------------

class StandardBlock(nn.Module):

    def __init__(self):

        super(StandardBlock, self).__init__()

        self.conv1 = nn.Conv2d(
            32,
            64,
            kernel_size=3,
            padding=1
        )

        self.conv2 = nn.Conv2d(
            64,
            32,
            kernel_size=3,
            padding=1
        )

        self.relu = nn.ReLU()

    def forward(self, x):

        identity = x

        out = self.conv1(x)

        out = self.relu(out)

        out = self.conv2(out)

        out = out + identity

        out = self.relu(out)

        return out


# ----------------------------------------------------------
# Step 4 - Create Bottleneck Block
# ----------------------------------------------------------

class BottleneckBlock(nn.Module):

    def __init__(self):

        super(BottleneckBlock, self).__init__()

        # Reduce number of channels
        self.reduce = nn.Conv2d(
            32,
            16,
            kernel_size=1
        )

        # Grouped convolution
        self.group_conv = nn.Conv2d(
            16,
            16,
            kernel_size=3,
            padding=1,
            groups=4
        )

        # Increase channels again
        self.expand = nn.Conv2d(
            16,
            32,
            kernel_size=1
        )

        self.relu = nn.ReLU()

    def forward(self, x):

        identity = x

        out = self.reduce(x)

        out = self.relu(out)

        out = self.group_conv(out)

        out = self.relu(out)

        out = self.expand(out)

        # Residual connection
        out = out + identity

        out = self.relu(out)

        return out


# ----------------------------------------------------------
# Step 5 - Create Models
# ----------------------------------------------------------

standard_model = StandardBlock()

bottleneck_model = BottleneckBlock()

print("Models Created Successfully")


# ----------------------------------------------------------
# Step 6 - Forward Pass
# ----------------------------------------------------------

standard_output = standard_model(X)

bottleneck_output = bottleneck_model(X)


print("\nStandard Block Output Shape:")
print(standard_output.shape)


print("\nBottleneck Block Output Shape:")
print(bottleneck_output.shape)


# ----------------------------------------------------------
# Step 7 - Count Parameters
# ----------------------------------------------------------

def count_parameters(model):

    total = 0

    for parameter in model.parameters():

        total += parameter.numel()

    return total


standard_parameters = count_parameters(
    standard_model
)

bottleneck_parameters = count_parameters(
    bottleneck_model
)


print("\nParameter Comparison")

print(
    "Standard Block:",
    standard_parameters
)

print(
    "Bottleneck Block:",
    bottleneck_parameters
)


# ----------------------------------------------------------
# Step 8 - Compare Parameter Reduction
# ----------------------------------------------------------

reduction = (
    1 -
    bottleneck_parameters /
    standard_parameters
) * 100


print(
    "\nParameter Reduction:",
    round(reduction, 2),
    "%"
)


# ----------------------------------------------------------
# Step 9 - Check Residual Connection
# ----------------------------------------------------------

print("\nResidual Connection Used:")
print("Input + Convolution Output")


# ----------------------------------------------------------
# Step 10 - Experiment Completed
# ----------------------------------------------------------

print("\nResNet Bottleneck Experiment Completed")
```
